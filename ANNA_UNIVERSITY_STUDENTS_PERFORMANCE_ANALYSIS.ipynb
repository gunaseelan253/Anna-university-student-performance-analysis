DEPARTMENT ANALYSIS

```python
department_count = df["Department"].value_counts()

print(department_count)
```

```python
plt.figure(figsize=(8,5))

department_count.plot(kind="bar")

plt.title("Students by Department")
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)
plt.show()
```

ACADAMIC PERFORMANCE ANALYSIS

```python
print("Average Mark:",
      df["Average_Mark"].mean())
```

```python
avg_marks = df.groupby("Department")["Average_Mark"].mean()

print(avg_marks)
```

```python
plt.figure(figsize=(8,5))

avg_marks.plot(kind="bar")

plt.title("Average Marks by Department")
plt.xlabel("Department")
plt.ylabel("Average Mark")
plt.xticks(rotation=45)
plt.show()
```

ATTENDANCE & BACKLOG ANALYSIS

```python
print("Average Attendance:",
      df["Attendance_Percentage"].mean())
```

```python
attendance = df.groupby("Department")["Attendance_Percentage"].mean()

print(attendance)
```

```python
print("Total Backlogs:",
      df["Backlogs"].sum())
```

```python
backlogs = df.groupby("Department")["Backlogs"].sum()

print(backlogs)
```

STUDENT SATISFACTION ANALYSIS

```python
print("Average Student Satisfaction:",
      df["Student_Satisfaction"].mean())
```

```python
satisfaction = df.groupby("Department")["Student_Satisfaction"].mean()

print(satisfaction)
```

```python
plt.figure(figsize=(8,5))

satisfaction.plot(kind="bar")

plt.title("Student Satisfaction by Department")
plt.xlabel("Department")
plt.ylabel("Satisfaction")
plt.xticks(rotation=45)
plt.show()
```

##SQL ANALYSIS

```python
!pip install duckdb -q
```

```python
import duckdb

con = duckdb.connect()

con.register("students", df)
```

TOTAL STUDENTS

```python
con.sql("""
SELECT COUNT(*) AS Total_Students
FROM students
""")
```

DEPARTMENT-WISE STUDENT

```python
con.sql("""
SELECT
    Department,
    COUNT(*) AS Student_Count
FROM students
GROUP BY Department
ORDER BY Student_Count DESC
""")
```

DEPARTMENT-WISE AVERAGE MARK

```python
con.sql("""
SELECT
    Department,
    ROUND(AVG(Average_Mark), 2) AS Average_Mark
FROM students
GROUP BY Department
ORDER BY Average_Mark DESC
""")
```

DEPARTMENT-WISE ATTENDANCE

```python
con.sql("""
SELECT
    Department,
    ROUND(AVG(Attendance_Percentage), 2) AS Average_Attendance
FROM students
GROUP BY Department
ORDER BY Average_Attendance DESC
""")
```

INTERSHIP ANALYSIS

```python
con.sql("""
SELECT
    Internship_Status,
    COUNT(*) AS Student_Count
FROM students
GROUP BY Internship_Status
""")
```
